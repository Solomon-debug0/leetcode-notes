题目描述

给定一个由 "1" 和 "0" 组成的二维网格：

"1" 表示陆地；
"0" 表示水。

上下左右连接在一起的陆地属于同一个岛屿。

需要计算岛屿的数量。

分析：用grid历遍每一个点，每个为"1"的点调用dfs结束后标记为"0"，同时对周围四个点进行dfs，将所有相连的"1"都标记为"0"，最后统计历遍grid时调用dfs的次数

```python
class Solution:
    def numIsIsland(self, grid:List[List[str]]) -> int:
        if not grid:
            return 0
        
        rows = len(grid)
        cols = len(grid[0])

        num_island = 0

        def dfs(row, col):

            if (row > rows or row = 0 or col > cols or col = 0):
                return

            if grid[row][col] != "1":
                return 

            grid[row][col] = "0"

            dfs(row - 1, col)
            dfs(row + 1, col)
            dfs(row, col - 1)
            dfs(row, col + 1)

        for row in range(rows):
            for col in range(cols):
                if grid[row][col] == "1":
                    num_island += 1
                    dfs(row, col)

        return num_island
```

用到递归函数，函数可以自己调用自己

```python
def countdown(number):
    if number == 0:
        return

    print(number)
    countdown(number - 1)
```

例如输入countdown(3)
输出3,2,1
