```python
from collections import deque
#更详细易懂的版本

class Solution:
    def maxSlidingWindow(self, nums, k):
        queue = deque()
        answer = []

        for right in range(len(nums)):
            left = right - k + 1

            # 删除过期下标
            while len(queue) > 0:
                first_index = queue[0]

                if first_index < left:
                    queue.popleft()
                else:
                    break

            # 删除比当前元素更小的队尾元素
            while len(queue) > 0:
                tail_index = queue[-1]
                tail_value = nums[tail_index]
                current_value = nums[right]

                if tail_value <= current_value:
                    queue.pop()
                else:
                    break

            queue.append(right)

            if right >= k - 1:
                max_index = queue[0]
                max_value = nums[max_index]
                answer.append(max_value)

        return answer
```
