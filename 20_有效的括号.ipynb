给定一个只包含以下字符的字符串：

( ) [ ] { }

判断括号是否合法。

合法需要满足：

左括号必须由相同类型的右括号闭合；
括号必须按照正确顺序闭合；
每个右括号都必须有对应的左括号。

分析：用列表模拟“栈”，注意：这里只有三种括号，不用考虑括号中有其他字符的情况。

```python
class Solutuon:
    def isvaild(self, s:str) ->bool:
        stark = []
        pairs = {
            ")": "(",
            "]": "[",
            "}": "{"
       }

        for ch in s:
            if ch in pairs:
                if not stark:
                    return False
                top = stark.pop()
                if top != pairs[ch]:
                    return False
                else:
                    stark.append(ch)
                    
        return not stark
```
