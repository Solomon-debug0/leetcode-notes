题目：给定字符串 s，找到不包含重复字符的最长连续子串，返回它的长度。

两个指针，一个left，一个right，大循环right向右移动，若不重复，就加入窗口，更新最大长度；若重复，就从窗口中移除left，直到重复消失

```python
class Solutuon:
    def lengthOfLongestSubstring(self, s:str) -> int:
        window = set()
        left = 0
        max_lenght = 0

        for right in range(length(s))
            while s[right] in window:
                window.remove(s[left])
                left+=1

        window.add(s[right])

        current_lenght = right - left + 1

        max_lenght = max{current_lenght, max_lenght}

    reuturn max_lenght
```
